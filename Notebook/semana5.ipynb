## Atividade 1
Escreva os métodos de ponto fixo dentro do arquivo $utils/algoritmos.py$.

````python
import numpy as np
from scipy.optimize import approx_fprime

def pontofixo(a, g, TOL=1e-8):
    x = g(a)
    while abs(x - a) > TOL:
        a = x
        x = g(a)
    return x

def newton_raphson(a, f, TOL=1e-8, df=None):
    if df is None:
        def dfn(x):
            return approx_fprime(np.array([x]), lambda v: f(v[0]))[0]
    else:
        dfn = df
    g = lambda x: x - f(x) / dfn(x)
    return pontofixo(a, g, TOL)

def secante(a, b, f, TOL=1e-8):
    g = lambda a, b: (a * f(b) - b * f(a)) / (f(b) - f(a))
    x = g(a, b)
    while abs(x - b) > TOL:
        a, b = b, x
        x = g(a, b)
    return x
````

Resolver a equação $e^x = x + 2$ é equivalente a calcular os pontos fixos da função  

$$
g(x) = e^x - 2
$$

Use os métodos do ponto fixo  

$$
x^{(n+1)} = g(x^{(n)})
$$

com $x^{(0)} = -1.8$ para obter uma aproximação de uma das soluções da equação dada com 8 dígitos significativos.  

**Resposta:**  
$
x \approx -1.8414057
$

````python
import numpy as np

import sys
import os

# Sobe um nível para a raiz do projeto e adiciona ao caminho do sistema
sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import (
    pontofixo,
    newton_raphson,
    secante,
)

f1 = lambda x: np.e**x - x - 2
g1 = lambda x: np.e**x - 2


def main():
    r = pontofixo(-1.8, g1)
    print(f"raiz ponto fixo = {r}")
    r = newton_raphson(-1.8, f1, df=lambda x: np.e**x - 1)
    print(f"raiz newton-raphson df = {r}")
    r = newton_raphson(-1.8, f1)
    print(f"raiz newton-raphson = {r}")
    r = secante(-1.8, -1.7, f1)
    print(f"raiz secante = {r}")


if __name__ == "__main__":
    main()
````

## Atividade 2
Encontre a raiz positiva da função  

$$
f(x) = \cos(x) - x^2
$$  

pelos métodos do ponto fixo, inicializando-o com $x^{(0)} = 1$.  

Realize a iteração até obter estabilidade no **quinto dígito significativo**.  

**Resposta:**

$$
x \approx 0.82413 
$$

Processo iterativo:  

$$
x^{(n+1)} = x^{(n)} + \frac{\cos(x) - x^2}{\sin(x) + 2x}
$$

````python
import numpy as np

def g(x):
    return x + (np.cos(x) - x**2) / (np.sin(x) + 2*x)

x = 1.0

tol = 1e-5
max_iter = 100

print(f"x_0 = {x:.6f}")

for i in range(1, max_iter + 1):
    x_next = g(x)
    print(f"x_{i} = {x_next:.6f}")

    if abs(x_next - x) < tol:
        print(f"\nConvergência atingida na iteração {i}!")
        print(f"Resultado com 5 dígitos significativos: {x_next:.5f}")
        break
        
    x = x_next
````

## Atividade 3
Aplique os métodos do ponto fixo para resolver a equação:

$$
e^{-x^2} = 2x
$$

**Resposta:**

$$
x \approx 0.4193648
$$

````python
import numpy as np

def g(x):
    return 0.5 * np.exp(-x**2)

x = 0.5

tol = 1e-7
max_iter = 100

for i in range(1, max_iter + 1):
    x_next = g(x)
    print(f"x_{i} = {x_next:.7f}")
    
    if abs(x_next - x) < tol:
        print(f"\nConvergência atingida na iteração {i}!")
        print(f"Resultado: {x_next:.7f}")
        break
        
    x = x_next
````

## Atividade 4
Resolva os três últimos exercícios da semana anterior usando os métodos do ponto fixo.

````python
import numpy as np #Atividade 4

k = 1.380649e-23
q = 1.602176634e-19
T = 300
vt = (k * T) / q 
I_R = 1e-12     

casos = [
    ("a", 30.0, 1000.0),
    ("b", 3.0, 1000.0),
    ("c", 3.0, 10000.0),
    ("d", 0.3, 1000.0),
    ("e", -0.3, 1000.0),
    ("f", -30.0, 1000.0),
    ("g", -30.0, 10000.0)
]

for item, V, R in casos:
    vd = V / 2.0 
    
    for _ in range(200):
        if V > 0:
            arg = 1.0 + (V - vd) / (R * I_R)
            vd_next = vt * np.log(max(arg, 1e-12))
        else:
            vd_next = V - R * I_R * (np.exp(vd / vt) - 1.0)
            
        if abs(vd_next - vd) < 1e-7:
            vd = vd_next
            break
        vd = vd_next
        
    print(f"Item {item}) vd = {vd:.3f} V")
````

````python
import numpy as np #Atividade 5

d = 500.0
f = 50.0

C = 600.0
tol = 1e-7

for i in range(100):
   
    C_next = (d / 2.0) / np.arccosh(1.0 + f / C)
    
    if abs(C_next - C) < tol:
        C = C_next
        break
    C = C_next

print(f"Parâmetro C = {C:.4f} m")

comprimento = 2 * C * np.sinh(d / (2 * C))
print(f"Comprimento do cabo (S) = {comprimento:.4f} m")
````

````python
import numpy as np #Atividade 6

f = 1000.0      
L = 100e-3      
R = 1000.0       


tan_phi = (2 * np.pi * f * L) / R
phi = np.arctan(tan_phi)

beta_rad = np.pi + phi  
tol = 1e-8

for _ in range(100):
    term = np.sin(phi) * np.exp(-beta_rad / tan_phi)
    beta_next = np.pi + phi - np.arcsin(term)
    
    if abs(beta_next - beta_rad) < tol:
        beta_rad = beta_next
        break
    beta_rad = beta_next

beta_deg = np.degrees(beta_rad)
print(f"Ângulo beta = {beta_deg:.4f}°")
````
