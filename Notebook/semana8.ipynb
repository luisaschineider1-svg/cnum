## Atividade 1
Para economizar energia elétrica, um agricultor implantou um sistema de painel solar na fazenda para alimentar uma bomba d’água, que faz a irrigação das plantações.  
A placa negra utilizada fica exposta ao sol e ao vento. Dessa forma, para gerar eletricidade, a irradiação solar $E$ (em $\text{W/m}^2$) sobre a placa tem que ser maior que a perda de calor por radiação e por condução, dada pela temperatura atmosférica $K$ (em Kelvin).

A equação é dada por:

$$
E = 5,67 \times 10^{-8} \, T^4 + 0,4(T - K)
$$

Determine a temperatura mínima da placa, dada a medição da irradiação e da temperatura do ar, sabendo que o valor médio diário é $E = 500,125$ e $K = 272,975$.

**Sugestão:** usar métodos da bissecção ou ponto fixo.  
**Resposta aproximada:**  

$$
T = 304{,}56801011987010952
$$

````python
E = 500.125
K = 272.975

def f(T):
    return 5.67e-8 * T**4 + 0.4 * (T - K) - E

# Intervalo inicial
a = 300.0
b = 310.0
tolerancia = 1e-12

if f(a) * f(b) > 0:
    raise ValueError("O intervalo não contém uma raiz.")

while (b - a) / 2 > tolerancia:
    meio = (a + b) / 2

    if f(meio) == 0:
        a = b = meio
        break
    elif f(a) * f(meio) < 0:
        b = meio
    else:
        a = meio

T = (a + b) / 2

print(f"Temperatura da placa: {T:.14f} K")
print(f"Resíduo na equação: {f(T):.12e}")
````

## Atividade 2
Em um grande hospital, para evitar intermitência no sistema elétrico dos diversos equipamentos médicos, instalou-se três reatores de alta performance em série, com tensão nominal $R$ em função da potência máxima $P$ de operação dos enrolamentos de controle:

$$
17R_1 - 2R_2 - 3R_3 = P_1 = 500
$$  

$$
-5R_1 + 21R_2 - 2R_3 = P_2 = 200
$$  

$$
-5R_1 - 5R_2 + 22R_3 = P_3 = 300
$$  

Calcular as tensões nominais de cada reator.

**Sugestão:** usar o método de Gauss-Seidel.  
**Resposta aproximada:**  

$$
R_1 = 36{,}56081655798128
$$  

$$
R_2 = 20{,}768358378225123
$$  

$$
R_3 = 26{,}66572157641055
$$  


````python
# Tensões iniciais
R1 = 0.0
R2 = 0.0
R3 = 0.0

tolerancia = 1e-12
max_iteracoes = 1000

for iteracao in range(1, max_iteracoes + 1):
    R1_antigo = R1
    R2_antigo = R2
    R3_antigo = R3

    # Atualizações de Gauss-Seidel
    R1 = (500 + 2 * R2 + 3 * R3) / 17
    R2 = (200 + 5 * R1 + 2 * R3) / 21
    R3 = (300 + 5 * R1 + 5 * R2) / 22

    maior_variacao = max(
        abs(R1 - R1_antigo),
        abs(R2 - R2_antigo),
        abs(R3 - R3_antigo)
    )

    if maior_variacao < tolerancia:
        break

print(f"R1 = {R1:.14f}")
print(f"R2 = {R2:.14f}")
print(f"R3 = {R3:.14f}")
print(f"Iterações: {iteracao}")
````

## Atividade 3
Determine o valor da corrente no resistor **R3**, sabendo que $V_1 = V_2 = 100\,V$ e $R_1 = R_2 = R_3 = 10\,\Omega$.

### Circuito equivalente

            R2           R1
     ┌─────/\/\/\───┬──/\/\/\───┐
     │             │            │
     │             │            │
    V2(+)        /\/\/\      V1(+)
     │            R3            │
     │             │            │
     └─────────────┴────────────┘

$$
\begin{bmatrix}
20 & 10 \\
10 & 20
\end{bmatrix}
\begin{Bmatrix}
I_1 \\
I_2
\end{Bmatrix}
=
\begin{Bmatrix}
100 \\
100
\end{Bmatrix}
$$

**Sugestão:** usar o método de Gauss-Seidel.

**Resposta aproximada:**  

$$
I_{R3} = 6{,}6667\,A
$$

````python
# Sistema:
# 20*I1 + 10*I2 = 100
# 10*I1 + 20*I2 = 100

I1 = 0.0
I2 = 0.0

tolerancia = 1e-10
max_iteracoes = 1000

for iteracao in range(1, max_iteracoes + 1):
    I1_anterior = I1
    I2_anterior = I2

    # Atualizações de Gauss-Seidel
    I1 = (100 - 10 * I2) / 20
    I2 = (100 - 10 * I1) / 20

    maior_variacao = max(
        abs(I1 - I1_anterior),
        abs(I2 - I2_anterior)
    )

    if maior_variacao < tolerancia:
        break
else:
    raise RuntimeError("O método não convergiu.")

corrente_R3 = I1 + I2

print(f"I1 = {I1:.6f} A")
print(f"I2 = {I2:.6f} A")
print(f"Corrente em R3 = {corrente_R3:.6f} A")
print(f"Iterações: {iteracao}")
````

## Atividade 4

Num sistema de energia solar, um vetor de equilíbrio energético $E$, na placa absorvente e na placa de vidro, segue o seguinte sistema de equações não lineares nas temperaturas absolutas $(K/m^2)$:

$$
(T_1^4 + 0,06823T_1) - (T_2^4 + 0,05848T_2) = E_1
$$    

$$
(T_1^4 + 0,05848T_1) - (2T_2^4 + 0,11696T_2) = E_2
$$  

Calcule o valor aproximado das temperaturas de equilíbrio, sabendo que os valores de equilíbrio energético foram medidos experimentalmente em:  

$E_1 = 0,01753$ e $E_2 = 0,00254$.

**Sugestão:** usar o método de Newton-Raphson.  
**Resposta aproximada:**  

$$
T_1 = 0,30543
$$  

$$ 
T_2 = 0,185261
$$ 

````python
E1 = 0.01753
E2 = 0.00254

# Chute inicial
T1 = 0.3
T2 = 0.18

tolerancia = 1e-12
max_iteracoes = 100

for iteracao in range(1, max_iteracoes + 1):
    # Sistema F(T1, T2) = 0
    F1 = T1**4 + 0.06823*T1 - T2**4 - 0.05848*T2 - E1
    F2 = T1**4 + 0.05848*T1 - 2*T2**4 - 0.11696*T2 - E2

    # Derivadas parciais (matriz Jacobiana)
    J11 = 4*T1**3 + 0.06823
    J12 = -(4*T2**3 + 0.05848)
    J21 = 4*T1**3 + 0.05848
    J22 = -8*T2**3 - 0.11696

    determinante = J11*J22 - J12*J21

    if abs(determinante) < 1e-15:
        raise ZeroDivisionError("A matriz Jacobiana é singular.")

    # Resolve J * delta = F
    delta_T1 = (F1*J22 - J12*F2) / determinante
    delta_T2 = (J11*F2 - J21*F1) / determinante

    # Atualização de Newton-Raphson
    T1_novo = T1 - delta_T1
    T2_novo = T2 - delta_T2

    variacao = max(abs(T1_novo - T1), abs(T2_novo - T2))
    T1, T2 = T1_novo, T2_novo

    if variacao < tolerancia:
        break
else:
    raise RuntimeError("O método não convergiu.")

print(f"T1 = {T1:.12f}")
print(f"T2 = {T2:.12f}")
print(f"Iterações: {iteracao}")
````
