## Atividade 1
Uma fábrica de laticínios, na etapa de resfriamento para fermentação dos produtos, utiliza diferentes tamanhos de serpentinas para alcançar as temperaturas ideais no processo de fabricação.  

De acordo com o manual do fabricante, as serpentinas seguem precisamente a fórmula:

$$
f(x) = a_3 x^3 + a_2 x^2 + a_1 x + a_0
$$

A tabela contém as temperaturas na primeira linha, os metros de serpentina na segunda, e o tamanho das novas serpentinas na terceira.  
Encontre as temperaturas alcançadas para essas novas serpentinas.

| Valores |
|----------|
| 4.9, 3.3, 3.0, 2.0 |
| 1.5, 2.0, 2.2, 3.0 |
| 1.75, 2.5, 2.75, 3.2 |
| **Resposta aproximada** |
| **3.91, 2.7, 2.43, 1.28** |

**Sugestão:** usar o método da **interpolação polinomial**.

````python
import sys, os
import numpy as np
sys.path.append(os.path.abspath(os.path.join('..')))
from utils.algoritmos import polinomial

y_temp = np.array([4.9, 3.3, 3.0, 2.0], dtype=float)
x_comp = np.array([1.5, 2.0, 2.2, 3.0], dtype=float)

x_novas = np.array([1.75, 2.5, 2.75, 3.2], dtype=float)

coef = polinomial(x_comp, y_temp)
temp_estimadas = np.polyval(coef, x_novas)

for comp, temp in zip(x_novas, temp_estimadas):
    print(f"Serpentina {comp:>4} m -> Temperatura estimada: {temp:.2f} °C")
````

## Atividade 2
Suponha um circuito RL em série com uma fonte de tensão constante $V$, resistência $R$ e indutância $L$.  
A corrente $I(t)$ é medida nos seguintes instantes:

| Tempo $t$ (s) | Corrente $I(t)$ (A) |
|----------------|---------------------|
| 0.2 | 0.8187 |
| 0.5 | 1.5815 |
| 0.7 | 1.9354 |
| 1.0 | 2.2905 |

Estime $I(0.3)$ e $I(0.9)$.

**Sugestão:** método da **interpolação polinomial**.  
**Resposta:**  
$I(0.3) = 1.107639\,A$  
$I(0.9) = 2.19182\,A$


````python
import sys, os
import numpy as np
sys.path.append(os.path.abspath(os.path.join('..')))
from utils.algoritmos import polinomial

t = np.array([0.2, 0.5, 0.7, 1.0], dtype=float)
I = np.array([0.8187, 1.5815, 1.9354, 2.2905], dtype=float)

coef = polinomial(t, I)
t_alvo = np.array([0.3, 0.9], dtype=float)
I_est = np.polyval(coef, t_alvo)

print(f"I(0.3) = {I_est[0]:.6f} A")
print(f"I(0.9) = {I_est[1]:.5f} A")
````

## Atividade 3

Deseja-se obter o modelo linear:

$$
f(x) = a_1 x + a_0
$$

onde $x$ é o comprimento da resistência (cm) e $f(x)$ é a temperatura (°C).

### Dados experimentais

| Modelo | Medições (°C) | Resistência (cm) | Resposta aproximada $(a_1, a_0)$ |
|---------|----------------|------------------|-----------------------------------|
| **Outono** | 38, 40, 42, 44, 46 | 1, 1.25, 1.5, 1.75, 2 | (30, 8) |
| **Inverno** | 40, 44, 48, 52, 56 | 1, 1.5, 2, 2.5, 3 | (32, 8) |
| **Primavera** | 36, 39, 42, 45, 48 | 1, 1.15, 1.3, 1.45, 1.6 | (16, 20) |

**Sugestão:** usar o método dos **quadrados mínimos**.

````python
import sys, os
import numpy as np
sys.path.append(os.path.abspath(os.path.join('..')))
from utils.algoritmos import regressao

v_lin = lambda x: np.column_stack((x, np.ones(len(x))))

dados = {
    "Outono":    (np.array([1.0, 1.25, 1.5, 1.75, 2.0]), np.array([38, 40, 42, 44, 46], dtype=float)),
    "Inverno":   (np.array([1.0, 1.5, 2.0, 2.5, 3.0]),   np.array([40, 44, 48, 52, 56], dtype=float)),
    "Primavera": (np.array([1.0, 1.15, 1.3, 1.45, 1.6]), np.array([36, 39, 42, 45, 48], dtype=float)),
}

for modo, (x, y) in dados.items():
    a1, a0 = regressao(x, y, v_lin)
    print(f"{modo:<10} -> f(x) = {a1:.2f}x + {a0:.2f}  | (a1, a0) = ({a1:.2f}, {a0:.2f})")
````

## Atividade 5

A relação entre tensão, corrente, resistência e indutância é:

$$
V(t) = L \frac{dI(t)}{dt} + R I(t)
$$

onde:  
$R = 5\,\Omega$, $L = 0.1\,H$

| Tempo $t$ (s) | Corrente $I(t)$ (A) |
|----------------|--------------------|
| 0.0 | 0.00 |
| 0.1 | 0.82 |
| 0.2 | 1.36 |
| 0.3 | 1.60 |
| 0.4 | 1.73 |

Sabendo que $I(t)$ cresce linearmente no tempo, estime $V(0.5)$.

**Sugestão:** método da **diferença finita progressiva**.  
**Resposta:** $V(0.5) = 12.29397\,V$

````python
import sys, os
import numpy as np
sys.path.append(os.path.abspath(os.path.join('..')))
from utils.algoritmos import regressao, dp

R, L = 5.0, 0.1
t = np.array([0.0, 0.1, 0.2, 0.3, 0.4], dtype=float)
I = np.array([0.00, 0.82, 1.36, 1.60, 1.73], dtype=float)

v_lin = lambda x: np.column_stack((x, np.ones(len(x))))
a1, a0 = regressao(t, I, v_lin)
I_lin = lambda x: a1 * x + a0

h = 1e-4
dI_dt = dp(I_lin, 0.5, h)
V_05 = L * dI_dt + R * I_lin(0.5)

print(f"V(0.5) = {V_05:.5f} V")
````

## Atividade 6

Repita a tarefa anterior, supondo que $I(t)$ cresce quadraticamente no tempo.

**Sugestão:** método da **diferença finita regressiva**.  
**Resposta:** $V(0.5) = 7.37401\,V$

````python
import sys, os
import numpy as np
sys.path.append(os.path.abspath(os.path.join('..')))
from utils.algoritmos import regressao, dr

R, L = 5.0, 0.1
t = np.array([0.0, 0.1, 0.2, 0.3, 0.4], dtype=float)
I = np.array([0.00, 0.82, 1.36, 1.60, 1.73], dtype=float)

v_quad = lambda x: np.column_stack((x**2, x, np.ones(len(x))))
a2, a1, a0 = regressao(t, I, v_quad)
I_quad = lambda x: a2 * (x**2) + a1 * x + a0

h = 1e-4
dI_dt = dr(I_quad, 0.5, h)
V_05 = L * dI_dt + R * I_quad(0.5)

print(f"V(0.5) = {V_05:.5f} V")
````

## Atividade 7

Repita novamente, considerando $I(t)$ cúbico no tempo.

**Sugestão:** método da **diferença finita central**.  
**Resposta:** $V(0.5) = 8.89841\,V$

````python
import sys, os
import numpy as np
sys.path.append(os.path.abspath(os.path.join('..')))
from utils.algoritmos import regressao, dc

R, L = 5.0, 0.1
t = np.array([0.0, 0.1, 0.2, 0.3, 0.4], dtype=float)
I = np.array([0.00, 0.82, 1.36, 1.60, 1.73], dtype=float)

v_cub = lambda x: np.column_stack((x**3, x**2, x, np.ones(len(x))))
a3, a2, a1, a0 = regressao(t, I, v_cub)
I_cub = lambda x: a3 * (x**3) + a2 * (x**2) + a1 * x + a0

h = 1e-4
dI_dt = dc(I_cub, 0.5, h)
V_05 = L * dI_dt + R * I_cub(0.5)

print(f"V(0.5) = {V_05:.5f} V")
````

## Atividade 8

Uma empresa de saneamento pública realiza o cálculo de micro drenagem das ruas por meio da pluviosidade das águas de chuva em função do raio da manilha, dado pela fórmula $r$ é:

$$
P(r) = 3\left(1 - \frac{r}{4}\right)^{1/7}
$$

A vazão de água de chuva é dada por:

$$
Q(r) = 2\pi \int_0^r r P(r)\,dr
$$

| Raio $r$ | Resposta $Q(r)$ |
|-----------|----------------|
| 1.25 | 14.235493 |
| 2.55 | 56.396741 |
| 3.15 | 83.297811 |
| 3.95 | 121.398311 |

**Sugestão:** usar a **regra de quadratura pelo ponto médio**.

````python
import sys, os
import numpy as np
sys.path.append(os.path.abspath(os.path.join('..')))
from utils.algoritmos import medio, integral

P = lambda r: 3.0 * ((1.0 - r / 4.0) ** (1.0 / 7.0))
integrando = lambda r: r * P(r)

raios = [1.25, 2.55, 3.15, 3.95]

print("Método do Ponto Médio:")
for r in raios:
    Q = 2 * np.pi * integral(medio, integrando, 0.0, r)
    print(f"r = {r:<5} -> Q(r) = {Q:.6f}")
````
